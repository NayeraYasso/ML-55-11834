# Task: Linear vs Polynomial Regression

## Objective
Build and compare Linear Regression and Polynomial Regression models then select the best performing model and interpret its learned equation.


## Dataset
Use the **Boston Housing dataset**.
    
    from sklearn.datasets import load_boston

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_absolute_error, mean_squared_error
from sklearn import metrics
```

```python
try:
    from sklearn.datasets import load_boston
    boston = load_boston()
    X = boston.data
    y = boston.target
    feature_names = boston.feature_names

except:
    from sklearn.datasets import fetch_openml
    boston = fetch_openml(name="boston", version=1, as_frame=True)
    X = boston.data
    y = boston.target
    feature_names = X.columns

X = pd.DataFrame(X, columns=feature_names)
y = pd.Series(y)
```

### 1. Data Preparation

```python
X = X.apply(pd.to_numeric, errors="coerce")
y = pd.to_numeric(y, errors="coerce")

data = X.copy()
data["target"] = y
data = data.dropna()

X = data.drop("target", axis=1)
y = data["target"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=101
)
```

### 2. Linear Regression Model
- Train a Linear Regression model
- Make predictions on the test set
- Compute evaluation metrics:
  - MAE
  - MSE
  - RMSE

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)
```

```python
linear_predictions = linear_model.predict(X_test)
```

```python
print("Linear Regression Results:")
print("MAE:", metrics.mean_absolute_error(y_test, linear_predictions))
print("MSE:", metrics.mean_squared_error(y_test, linear_predictions))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, linear_predictions)))
```

### 3. Polynomial Regression Model
- Train a Polynomial Regression model
- Make predictions on the test set
- Compute evaluation metrics:
  - MAE
  - MSE
  - RMSE


```python
poly = PolynomialFeatures(degree=2)

X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

```python
poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)
```

```python
poly_predictions = poly_model.predict(X_test_poly)
```

```python
print("Polynomial Regression Results:")
print("MAE:", metrics.mean_absolute_error(y_test, poly_predictions))
print("MSE:", metrics.mean_squared_error(y_test, poly_predictions))
print("RMSE:", np.sqrt(metrics.mean_squared_error(y_test, poly_predictions)))
```

### 4. Model Comparison
- Compare Linear vs Polynomial Regression results
- Identify which model performs better on the test data and justify your decision


```python
linear_rmse = np.sqrt(metrics.mean_squared_error(y_test, linear_predictions))
poly_rmse = np.sqrt(metrics.mean_squared_error(y_test, poly_predictions))

print("Linear RMSE:", linear_rmse)
print("Polynomial RMSE:", poly_rmse)

if poly_rmse < linear_rmse:
    print("Polynomial Regression performs better.")
else:
    print("Linear Regression performs better.")
```
